```python
#%% Import packages
import pandas as pd
import numpy as np
import numpy as np
from numba import njit
import matplotlib.pyplot as plt
from math import gamma
from scipy.optimize import fmin
from scipy.optimize import differential_evolution

import warnings
warnings.filterwarnings('ignore')
```

```python
# ========================================================================================================
# Parameters
# ========================================================================================================

class structtype():
    pass

p = structtype()

# Preference parameters
p.β  = 0.9675                                       # subjective time discounting
p.γ  = 6.00                                        # risk aversion
p.ψ  = 1.50                                        # elasticity of intertemporal substitution
p.α  = (1-p.γ)/(1-(1/p.ψ))                         # EZ composite parameter

# Production process
# p.Δθ  = 0.00
# p.Δτd = 0.00
# p.Δτ  = 0.00
# p.Δν  = 0.00
# p.Hf  = (35/79)
p.Δθ  = 0.045
p.Δτd = 0.10
p.Δτ  = 0.048
p.Δν  = 0.058
p.Hf  = 0.82
p.δ   = 0.07                                       # proportion of Elites
p.q   = (35/79)
p.θA  = 0.456 + p.δ                                # inequality parameter
p.θDH = p.θA - (p.Hf/p.q)*p.Δθ                     # inequality parameter
p.θDL = p.θA - ((1-p.Hf)/(1-p.q))*p.Δθ             # inequality parameter
p.yA  = 0.014                                      # drift of income in autocracy
p.ΔyD = 0.000                                      # change in income growth in democracy
p.yD  = p.yA + p.ΔyD                               # drift of income in democracy
p.σy  = 0.073                                      # volatility of income
p.νA  = 0.354                                      # Portion of consumption from public goods
p.νDH = (p.νA - (p.Hf/p.q)*(p.Δν))                 # Portion of consumption from public goods
p.νDL = (p.νA - ((1-p.Hf)/(1-p.q))*(p.Δν))         # Portion of consumption from public goods

# Tax process
p.τA   = 0.175
p.τH   = p.τA + (p.Hf/p.q)*p.Δτ
p.τL   = p.τA + ((1-p.Hf)/(1-p.q))*p.Δτ
p.ωH   = (p.θDH-p.δ)/(p.τH*(1-p.δ))
p.ωL   = (p.θDL-p.δ)/(p.τL*(1-p.δ))
# print((p.θDH-p.δ)/(p.ω*(1-p.δ)))

# Transition matrix
demL = 8.5
p.Pμ = np.array([[0.99,0.01,0.000],    # transition matrix for mu
                [1/(2*demL),1-1/demL,1/(2*demL)],
                [0.000,0.000,1.000]])


# Dividend process
p.ξ   = 0.77                                        # cross-sectional risk in democratization
p.Υ   = 2.6                                         # Leverage on growth and SD
```

```python
# ========================================================================================================
# Create functions for later use
# ========================================================================================================

# Felicity(ish) function
def u(c):
    return (1-β)*c**(1-1/ψ)

# After-tax income
def ŷ(τ,s,sg,ω):
    return (1-τ)*s + (sg*τ-0.5*ω*τ**2)
    # return (1-τ)*s + (τ)

# Disaster outcome
def χ(power):
    χTH = (ŷDrH/ŷAr)**power
    χTL = (ŷDrL/ŷAr)**power
    hg  = np.exp(power*ΔyD)
    return ( q*χTH + (1-q)*χTL )


# Cashflow drop for incumbents
def χD(power):
    ZDH = 1-(Hf/q)*(1-((1-(.2+Δτd))/(1-.2))*ξ)
    ZDL = 1-((1-Hf)/(1-q))*(1-((1-(.2+Δτd))/(1-.2))*ξ)
    χTH = (ŷDrH/ŷAr)**power
    χTL = (ŷDrL/ŷAr)**power
    return ( q*χTH*ZDH + (1-q)*χTL*ZDL )

# Consumption-wealth ratio
def consumptionWealth(κ0):

    χg = np.array([1,1,χ(1-γ)])
    Δ = np.array([
                  κ0[0] - 1 - βsA*(Pμ[0,:] @ ((κ0**α)*χg))**(1/α),
                  κ0[1] - 1 - βsA*(Pμ[1,:] @ ((κ0**α)*χg))**(1/α),
                  κ0[2] - 1 - βsD*(Pμ[2,:] @ ((κ0**α)))**(1/α)
    ])

    return(Δ.T @ Δ)

def priceDividend(pd0):

    βA = (β**α)*np.exp((Υ-γ)*yA + .5*((Υ-γ)**2)*σy**2)       # Adjusted beta in democracy
    βD = (β**α)*np.exp((Υ-γ)*yD + .5*((Υ-γ)**2)*σy**2)       # Adjusted beta in democracy

    pd1, pd2, pd3 = pd0[0],pd0[1],pd0[2]
    χgD = np.array([1,1,χD(-γ)])

    Δ = np.array([
                  pd0[0] - (βA * Pμ[0,:] @ ((κ/(κ[0]-1))**(α-1) * χgD * (pd0+1) ) ),
                  pd0[1] - (βA * Pμ[1,:] @ ((κ/(κ[1]-1))**(α-1) * χgD * (pd0+1) ) ),
                  pd0[2] - (βD * Pμ[2,:] @ ((κ/(κ[2]-1))**(α-1) * (pd0+1) ) )
    ])

    return Δ.T @ Δ
```

```python
# ========================================================================================================
# Unpack parameters and felicity function
# ========================================================================================================

# Unpack preference parameters
β, γ, ψ, α  = p.β, p.γ, p.ψ, p.α

# Unpack production process
θA, θDH, θDL, δ  = p.θA, p.θDH, p.θDL, p.δ
yA, ΔyD, σy, ξ, νDH, νDL, νA, ωH, ωL  = p.yA, p.ΔyD, p.σy, p.ξ, p.νDH, p.νDL, p.νA, p.ωH, p.ωL
τH, τL, τA, q, Hf, Δτd = p.τH, p.τL, p.τA, p.q, p.Hf, p.Δτd
yD = yA + ΔyD
Υ  = p.Υ

# Unpack grid and transition matrix for μ and τ
Pμ = p.Pμ

βsA   = β*np.exp((1-1/ψ)*yA + .5*(1-γ)*(1-1/ψ)*σy**2)     # Adjusted beta in autocracy
βsD   = β*np.exp((1-1/ψ)*yD + .5*(1-γ)*(1-1/ψ)*σy**2)     # Adjusted beta in democracy
spA   = (1-θA)/(1-δ)                                      # Citizen share of income
spD   = (1-θDH)/(1-δ)                                     # Citizen share of income
srA   = θA/δ                                              # Elite share of income in autocracy
srDH  = θDH/δ                                             # Elite share of income in democracy
srDL  = θDL/δ                                             # Elite share of income in democracy
sgrDH = νDH/δ
sgrDL = νDL/δ
sgrA  = νA/δ
sgp   = (1-δ)/(1-δ)

# Different income processes in different states
ŷDrH = ŷ(τH,srDH,sgrDH,ωH)
ŷDrL = ŷ(τL,srDL,sgrDL,ωL)
ŷAr  =  ŷ(τA,srA,sgrA,ωL)
ŷDp  =  ŷ(τH,spD,sgp,ωL)
ŷAp  = ŷ(τA,spA,sgp,ωL)
Δcr  = 1 - q*(ŷDrH/ŷAr) - (1-q)*(ŷDrL/ŷAr)

# Grid of y and c
yrgridL = np.array([ŷAr,ŷAr,ŷDrL])
yrgridH = np.array([ŷAr,ŷAr,ŷDrH])
ypgrid  = np.array([ŷAp,ŷAp,ŷDp])

# Value function in democracy and revolution
vDrH = (((1-β)/(1-βsD))**(1/(1-1/ψ)))*ŷDrH
vDrL = (((1-β)/(1-βsD))**(1/(1-1/ψ)))*ŷDrL
```

```python
def vSolve(vT,Zl):
    
#     print(vT[0],Zl)
    vDF = vD*(1-Zl[0])
    vgrid = np.array([vDF,vT[0],vA])

    Δ = np.array([
                  vT**(1-1/ψ) - u(ŷDr) - βsA * ( Qg @ vgrid**(1-γ) ) ** ((1-1/ψ)/(1-γ))
                 ])
                 
    return Δ.T @ Δ

def findZl(Zl):
    vT = fmin(func   = vSolve,
              x0     = 6,
              ftol   = 1**(-10),
              args   = (Zl,),
              disp   = False
             )
    return abs(vT - vD)

# Disaster outcome
def χ(power):
    χF = ((ŷDr*(1-Zl))/ŷDr)**power
    χS = (ŷAr/ŷDr)**power
    χgrid = np.array([χF,1,χS])

    return χgrid


# Consumption-wealth ratio
def consumptionWealth(κ0):

    χg = χ(1-γ)
    Δ = np.array([
                  κ0[0] - 1 - βsA*((κ0[0]**α))**(1/α),
                  κ0[1] - 1 - βsA*(Qg @ ((κ0**α)*χg))**(1/α),
                  κ0[2] - 1 - βsA*((κ0[2]**α))**(1/α),
    ])

    return(Δ.T @ Δ)

def priceDividend(pd0):

    βA = (β**α)*np.exp((Υ-γ)*yA + .5*((Υ-γ)**2)*σy**2)       # Adjusted beta in democracy
    # βD = (β**α)*np.exp((Υ-γ)*yD + .5*((Υ-γ)**2)*σy**2)       # Adjusted beta in democracy

    pd1, pd2, pd3 = pd0[0],pd0[1],pd0[2]
    χg = χ(Υ-γ)

    Δ = np.array([
                  pd0[0] - (βA * ((κ[0]/(κ[0]-1))**(α-1) * (pd0[0]+1) ) ),
                  pd0[1] - (βA * Qg @ ((κ/(κ[1]-1))**(α-1) * χg * (pd0+1) ) ),
                  pd0[2] - (βA * ((κ[2]/(κ[2]-1))**(α-1) * (pd0[2]+1) ) )
    ])

    return Δ.T @ Δ

```

```python
ŷDr  = ŷDrH/2 + ŷDrL/2
vA   = (((1-β)/(1-βsA))**(1/(1-1/ψ)))*ŷAr
vD   = (((1-β)/(1-βsD))**(1/(1-1/ψ)))*ŷDr
Qg   = np.array([.1,.8,.1])

Zl = fmin(func   = findZl,
          x0     = .5,
          ftol   = 1**(-10),
          disp   = False
         )[0]

κ = fmin(func   = consumptionWealth,
         x0     = [20,20,25],
         ftol   = 0.00000001,
         disp   = True
        )

### Run optimizer
pd = fmin(func   = priceDividend,
                x0     = [20,20,25],
                ftol   = 1**(-10),
                disp   = True
               )

Zl_base   = Zl
cg_base   = ŷAr/ŷDr
D_dp_base = np.log(1/pd[1]) - np.log(1/pd[0])
```
